## 4. Параметры (правьте здесь)

Все настраиваемые параметры сборки.

```python
from config.default_config import (
    Config, VideoConfig, SubtitleConfig, AudioConfig, TransitionConfig,
)

config = Config(
    project_path="input/Ролик",
    output_path="output/Ролик.mp4",
    video=VideoConfig(
        quality="720p",        # 360p / 480p / 720p / 1080p / 2k / 4k
        fps=30,                 # частота кадров
        codec="libx264",
        output_format="mp4",
    ),
    subtitle=SubtitleConfig(
        font_path="arial.ttf",          # путь к шрифту
        font_size=48,                    # базовый размер (для 1080p, масштабируется)
        font_color=(255, 255, 255),      # цвет текста RGB
        position="bottom",              # bottom / top / center
        margin_left_pct=5.0,             # отступ слева, % от ширины
        margin_right_pct=5.0,            # отступ справа, % от ширины
        margin_bottom_pct=8.0,           # отступ снизу, % от высоты
        margin_top_pct=8.0,              # отступ сверху, % от высоты
        max_lines=3,                     # макс. строк при переносе
        background_color=(0, 0, 0, 160), # подложка RGBA
        background_padding=12,
    ),
    audio=AudioConfig(
        normalize=False,     # выравнивание громкости: False/True
        gain_percent=100.0,  # 100=без изменений, 120=+20%, 80=-20%
    ),
    transition=TransitionConfig(
        type="crossfade",    # none / fade / crossfade / slide_left
        duration=0.5,
    ),
)
print("Конфигурация загружена:", config.to_dict())
```
